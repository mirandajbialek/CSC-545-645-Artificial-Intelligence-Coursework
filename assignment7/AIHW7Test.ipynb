```python
import random
def trial():
    count = 0
    money = 10
    while money >= 1:
        money -=1
        count += 1
        slots = [random.choice(["bar", "bell", "lemon", "cherry"]) for i in range(3)]
        if slots[0] == slots[1]:
            if slots[1] == slots[2]:
                num_equal = 3
            else:
                num_equal = 2
        else:
            num_equal = 1
        if slots[0] == "cherry":
            money += num_equal 
        elif num_equal == 3:
            if slots[0] == "bar":
                money += 20
            elif slots[0] == "bell":
                money += 15
            else:
                money += 5
    return count
```

```python
def test(trials):
    results = [trial() for i in range(trials)]
    mean = sum(results) / trials
    sorted_results = sorted(results)
    if trials % 2 == 0:
        median = (sorted_results[trials // 2 - 1] + sorted_results[trials // 2]) / 2.0
    else:
        median = sorted_results[trials // 2]
    print("%s trials, mean=%s, median=%s" % (trials, mean, median))
```

```python
test(1000)
```
